
# Basic instructions to Databricks Notebooks

Databricks notebooks are easy way to execute Apache Spark jobs in cloud environments with browser.
Here are a few pointers on how to get started.

For more information, see [official instructions](https://learn.microsoft.com/en-gb/azure/databricks/notebooks/notebooks-code).

### Going around in the workspace
* Your code is, on default, stored in the Databricks Workspace path. You can get there by selecting "Workspace" in the left menu and then Workspace / Users / Your email.
* The workspace will be destroyed at the end of the course. Thus, all your code will eventually disappear => make sure to backup your work!
* The recommended way is to use Databricks git repository support. The instructions for integrating a GitLab repository to Databricks can be found in the course Moodle in the instructions section.
    * Integrated Git repositories are shown as folders (marked with branch information) in the Databricks workspace.

### Executing code
* To do anything useful, you need to attach your notebook to a computational cluster. You can do this with the top-right "Connect" button in the notebook.
* If nobody else is using the cluster, the cluster might start when you connect to it. This might take ~5 minutes.
* All students are using the same computational cluster.
* Notebooks are divided to cells. You can execute cell code by clicking top right corner play button or by using "Shift"+"Enter" shortcut.
    * The "Run all" button tries to execute all cells in order from top to bottom. It will stop at the first encountered error.
    * The initiated variables will remain in the notebook "memory" as long as the notebook is connected to a computational cluster. \
      To run the notebook from a "fresh" state, Use the menu in top-left: "Run" / "Clear" / "Clear state and run all"
* You can see the other keyboard shortcuts via "Help" / "Keyboard shortcuts"
* Apache Spark cluster consists of Driver and Worker nodes. All Spark commands are shared between workers, all "pure" Scala and Python code is executed on Driver node. \
  Thus, try to avoid running executing heavy loads without Spark.

### Selecting programming language
* Each notebook has a default language. You can select it on the top of the notebook. In this notebook, the default language is "Scala"
* Cells can have different languages. However, you cannot directly refer to variables in other languages.
* You can select the cell-specific language from the top right corner of the cell. Alternatives are Scala, Python, and Markdown (for documentation). Databricks also supports R and SQL cells, but they are not allowed in course assignments.
* You can also use "magic commands" on the first line of cell to set the language. The magic commands are `%scala`, `%python`and `%md`.

### Working with cells
* You can add and remove cells, and change their order.
* You can also execute cells in any order.
* This means, you might end up with a very confusing ordering of cells. And notebooks that do not actually work when executing from the beginning to end.
* That is, your assignment might not work for us if you are not careful. Remember to try execution in the correct order.
* Using functional programming practices, i.e. immutable variables, helps to avoid problems related to cell order.

### Important
* Before submitting your code, **always execute notebook fully with clean state**. This is done by selecting "Run" / "Clear" / "Clear state and run all".
* Due to permission restrictions, every student can restart the cluster. **Do NOT restart clusters unnecessarily**. \
  Otherwise everyone's notebook executions will be cleared. However, any code will not disappear (but you will likely annoy others working at the same time).

```scala
// Let us execute Scala
// Because Scala is the default notebook language, we do not need the magic command %scala on the first line
val string_list : List[String] = List("This", "is", "Scala")
val printed_string = string_list.mkString(" ")
println("** Let's start our first printing.")
println(printed_string + " code!")
println("** Successfully printed our first text. Rest is other cell output.")
```

**Then some example Spark code.**

This Markdown cell is created with magic command. Double click this cell to see the magic command `%md` on the first line.

```scala
// Let us work on Scala and Apache Spark

// Create a sample DataFrame
val data = Seq(
  ("Alice", 25),
  ("Bob", 30),
  ("Charlie", 35)
)

// Let us do our first Apache Spark call. You can recognize it on the next line because we use library "spark"
val df = spark.createDataFrame(data).toDF("Name", "Age")
// Above code is now be divided between worker nodes

// Use Apache Spark print to show the result
df.show()
```

```scala
// However, Databricks also has a way to print DataFrames in nicer format
display(df)
```

```scala
// Let us do a bit of computation with Spark
import org.apache.spark.sql.functions._ // Very often DataFrame computations need this import

val avgAge = df.select(avg("Age")).head().getDouble(0)
val maxAge = df.select(max("Age")).head().getInt(0)
println(s"Average Age: $avgAge")
println(s"Maximum Age: $maxAge")
```

Now, let's go through some examples on how to read the data int the storage with Databricks Apache Spark. The address for the data is
`wasbs://<container>@tunics320f2026gen2.blob.core.windows.net/<path>/<to>/<file.csv>`

```scala
val file_csv = "wasbs://shared@tunics320f2026gen2.blob.core.windows.net/demo/kaggle/csv/10mb_imdb_anime.csv"

val df_csv = spark.read
  .option("header", "true")       // The first row has column names
  .option("sep", ",")             // "," is used as the column separator in the CSV file
  .option("inferSchema", "true")  // Try to automatically determine the data types for the columns
  .csv(file_csv)

display(df_csv)
```

```scala
%python
# The same with Python
file = "wasbs://shared@tunics320f2026gen2.blob.core.windows.net/demo/kaggle/csv/10mb_imdb_anime.csv"
df = spark.read  \
  .option("header", "true") \
  .option("sep", ",") \
  .option("inferSchema", "true") \
  .csv(file)

display(df)
```

```scala
// Typically we use some more suitable file format, like Parquet.
// Column format is stored in the file itself, so we do not need to give it.
val file_parquet = "wasbs://shared@tunics320f2026gen2.blob.core.windows.net/demo/kaggle/parquet/10mb_imdb_anime.parquet"
val df_parquet = spark.read.parquet(file_parquet)
display(df_parquet)
```

```scala
// There is also a possibility to use a more advanced Delta file format, also called "Delta lake table"
// We will cover this later in the course
val file_delta = "wasbs://shared@tunics320f2026gen2.blob.core.windows.net/demo/kaggle/delta/10mb_imdb_anime_delta"
val df_delta = spark.read.format("delta").load(file_delta)
display(df_delta)
```

# Writing files

Mostly during the course we will be just reading files in this course. However, if you want, you can also try writing the files.

```scala
val student_name = "example_student"  // Change this to your own name or some other unique identifier
```

```scala
val target_path = s"wasbs://students@tunics320f2026gen2.blob.core.windows.net/${student_name}/demo/"
val target_file = target_path + "10mb_imdb_anime.parquet"

// Write to Parquet
df_parquet.write
    .mode("overwrite")
    .parquet(target_file)
```

```scala
// Use Databricks specific utilities library to see the files. Spark has Hadoop filesystem library for the same, but it is not as simple
display(dbutils.fs.ls(target_path))
```

```scala
// Actually Parquet file is usually a folder with multiple files. And even CSV files are written to a folder.
// Workers might write to multiple files at the same time. Let's check the actual files
display(dbutils.fs.ls(target_file))
```
