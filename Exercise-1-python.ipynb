## Task 3 - Fibonacci numbers

The Fibonacci numbers, `F_n`, are defined such that each number is the sum of the two preceding numbers. The first two Fibonacci numbers are:

$$F_0 = 0 \qquad F_1 = 1$$

In the following cell, write a **recursive** function, `fibonacci`, that takes in the index and returns the Fibonacci number. (no need for any optimized solution here)

```python
def fibonacci(n):
    if n <= 0:
        return 0
    elif n == 1:
        return 1
    
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
print(f"- fibonacci(1) = {fibonacci(1)}")
print(f"- fibonacci(5) = {fibonacci(5)}")
print(f"- fibonacci(9) = {fibonacci(9)}")
print(f"- fibonacci(12) = {fibonacci(12)}")
print(f"- fibonacci(17) = {fibonacci(17)}")
```

```python
from functools import reduce

myList: list[int] = [2, 3, 5, 7, 11, 13, 17, 19, 21, 23, 29, 31, 37, 41, 43]
myList = map(lambda x: x**3, myList)

cubeSum: int =  reduce(lambda x,y: x+y ,myList)

myList: list[int] = [2, 3, 5, 7, 11, 13, 17, 19, 21, 23, 29, 31, 37, 41, 43]
vals = map(lambda x: x**2 - 34*x +325, myList)
smallestFuncValue: int = reduce(lambda a, b: a if a<b else b, vals)
```
